```python
print("00_joke_bot.")

PROMPT:str ="What do you want?"
JOKE:str = "Why dont skeletons fight each other? \n Because they don't have the guts!😊"
SORRY:str = " Sorry!I only tell jokes"

def main():
  user_input = input(PROMPT)
  user_input = user_input.strip().lower()
  #while user_input != "quit":
  if "joke" in user_input:
      print(JOKE)
  else:
      print(SORRY)

if __name__ == "__main__":
  main()


```

```python
print("01_double_it")
def main():
  user_value = int(input("Please enter a number: "))
  while user_value < 100:
    user_value *= 2
  print(user_value * 2)
if __name__ == "__main__":
  main()
```

```python
print("02_liftoff.")

def main():
  for i in range(10, 0, -1):
    print(i, end = "")
  print("Liftoff!")

if __name__ == "__main__":
  main()
```

```python
import random

print("03_guess_my_number")

def main():
    secret_number = random.randint(1, 100)
    print("I'm thinkiing of a number between 1 and 100.")

    guess = int(input("Enter a guess: "))

    while guess != secret_number:
        if guess < secret_number:
            print("Too low.")
        else:
            print("Too high.")

        guess = int(input("Enter a guess: "))  # Moved this outside the if-else block

    print(f"Congratulations! The number was {secret_number}")

if __name__ == "__main__":
    main()
```
