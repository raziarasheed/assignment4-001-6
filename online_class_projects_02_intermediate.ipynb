```python
import random

print("High Low Game")

rounds = 5  # Renamed from 'round' to 'rounds' to avoid conflict with built-in 'round'

def main():
    print("Welcome to the High Low Game")
    print("*********************************")

    score = 0
    for i in range(rounds):
        print("Round", i + 1)  # Fixed round counter
        computer_number = random.randint(1, 100)
        your_number = random.randint(1, 100)
        print("Your number is", your_number)

        choice = input("Do you think your number is higher or lower than the computer's number? (Type 'higher' or 'lower'): ").strip().lower()

        higher_and_correct = choice == "higher" and your_number > computer_number
        lower_and_correct = choice == "lower" and your_number < computer_number

        if higher_and_correct or lower_and_correct:
            print("You were right! The computer's number was", computer_number)
            score += 1
        else:
            print("That's incorrect. The computer's number was", computer_number)

        print()

    print("Your final score is", score)
    print("Thanks for playing!")

if __name__ == "__main__":
    main()
```

```python
print("Lists and Dicts")

# Sample list
my_list = ["apple", "mango", "orange", "pear", "peach"]

def access_element(my_list, index):
    """Returns the element at the specified index, or an error message if out of range."""
    if 0 <= index < len(my_list):
        return f'Element at index {index}: {my_list[index]}'
    else:
        return f'Error: Index {index} is out of range'

def modify_element(my_list, index, new_value):
    """Modifies the element at the specified index with the new value."""
    if 0 <= index < len(my_list):
        old_value = my_list[index]
        my_list[index] = new_value
        return f'Element at index {index} modified from {old_value} to {new_value}'
    else:
        return f'Error: Index {index} is out of range'

def slice_list(my_list, start_index, end_index):
    """Returns a new list containing the elements from the start index to the end index (exclusive)."""
    if 0 <= start_index < len(my_list) and 0 <= end_index <= len(my_list):
        return f'Sliced list: {my_list[start_index:end_index]}'
    return "Invalid slice indices!"

def main():
    print("\nWelcome to the List Manipulation Game!")

    while True:
        print("\nSelect an option:")
        print("1. Access Element")
        print("2. Modify Element")
        print("3. Slice List")
        print("4. Exit")

        choice = input("Enter your choice (1-4): ")

        if choice == "1":
            try:
                index = int(input("Enter the index of the element to access: "))
                print(access_element(my_list, index))
            except ValueError:
                print("Please enter a valid integer.")

        elif choice == "2":
            try:
                index = int(input("Enter the index of the element to modify: "))
                new_value = input("Enter the new value: ")
                print(modify_element(my_list, index, new_value))
            except ValueError:
                print("Please enter a valid integer.")

        elif choice == "3":
            try:
                start_index = int(input("Enter the start index of the slice: "))
                end_index = int(input("Enter the end index of the slice: "))
                print(slice_list(my_list, start_index, end_index))
            except ValueError:
                print("Please enter valid integers.")

        elif choice == "4":
            print("Exiting the Game. Thanks for playing!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 4.")

# Entry point
if __name__ == "__main__":
    main()

#
```
